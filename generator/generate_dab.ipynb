```python
import json

workflow_file = "/Volumes/accelerator/metadata/generated_code/customer_workflow.json"

with open(workflow_file, "r") as f:
    workflow = json.load(f)

print(workflow)
```

```python
for task in workflow["tasks"]:
    print(task)
```

```python
tasks_yaml = ""

for task in workflow["tasks"]:

    task_key = task["task_key"]

    tasks_yaml += f"""
      - task_key: {task_key}
"""

    if task["depends_on"]:

        tasks_yaml += """
        depends_on:
"""

        for dep in task["depends_on"]:

            tasks_yaml += f"""
          - task_key: {dep}
"""

    tasks_yaml += f"""
        spark_python_task:
          python_file: ../src/{task_key}.py
"""
```

```python
workflow_yaml = f"""
resources:
  jobs:
    {workflow['workflow_name']}:

      name: {workflow['workflow_name']}

      tasks:
{tasks_yaml}
"""
```

```python
print(workflow_yaml)
```

```python
workflow_path = "/Volumes/accelerator/metadata/generated_code/bundle/resources/workflow.yml"

with open(workflow_path, "w") as f:
    f.write(workflow_yaml)

print("workflow.yml generated")
```
