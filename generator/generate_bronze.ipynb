```python
df = spark.table("accelerator.metadata.table_configs")

display(df)
```

```python
bronze_template = """
import dlt

@dlt.table(
    name="bronze_{table_name}"
)
def bronze_{table_name}():

    return (
        spark.readStream
             .format("cloudFiles")
             .option("cloudFiles.format", "csv")
             .load("{source_path}")
    )
"""
```

```python
for row in df.collect():

    code = bronze_template.format(
        table_name=row["table_name"],
        source_path=row["source_path"]
    )

    print(code)
```

```python
# path = f"/Volumes/accelerator/metadata/generated/bronze_{table}.py"
```

```python
for row in df.collect():

    table_name = row["table_name"]

    code = bronze_template.format(
        table_name=row["table_name"],
        source_path=row["source_path"]
    )

    file_path = f"/Volumes/accelerator/metadata/generated_code/bronze_{table_name}.py"

    with open(file_path, "w") as f:
        f.write(code)

    print(f"Generated: {file_path}")
```
