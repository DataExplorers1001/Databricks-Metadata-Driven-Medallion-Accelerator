```python
table_df = spark.table(
    "accelerator.metadata.table_configs"
)

workflow_df = spark.table(
    "accelerator.metadata.workflow_configs"
)
```

```python
import json

workflow = workflow_df.first()

workflow_name = workflow["workflow_name"]

tasks = []

for row in table_df.collect():

    table_name = row["table_name"]

    tasks.append({
        "task_key": f"bronze_{table_name}",
        "depends_on": []
    })

    tasks.append({
        "task_key": f"silver_{table_name}",
        "depends_on": [f"bronze_{table_name}"]
    })

    tasks.append({
        "task_key": f"gold_{table_name}",
        "depends_on": [f"silver_{table_name}"]
    })
```

```python
workflow_json = {
    "workflow_name": workflow_name,
    "tasks": tasks
}
```

```python
workflow_json
```

```python
file_path = "/Volumes/accelerator/metadata/generated_code/customer_workflow.json"

with open(file_path, "w") as f:
    json.dump(
        workflow_json,
        f,
        indent=4
    )
```
