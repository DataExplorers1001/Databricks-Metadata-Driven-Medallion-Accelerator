```python
from pyspark.sql.functions import current_date

drift_count = spark.sql("""
SELECT COUNT(*)
FROM accelerator.metadata.schema_drift_audit
WHERE CAST(detected_on AS DATE) = CURRENT_DATE()
""").collect()[0][0]

print(f"Drift Count = {drift_count}")
```

```python
if drift_count > 0:
    raise Exception(
        f"Schema Drift Detected. Count = {drift_count}"
    )
else :
    print("No Schema Drift Detected")
    # %run ./run_accelerator 
    dbutils.notebook.run("./run_accelerator ", 0)
```
