```python
df = spark.table("accelerator.metadata.table_configs")
```

```python
from pyspark.sql.functions import current_timestamp,expr

lineage_df = (
    df.select("table_name")
      .withColumn(
          "bronze_table",
          expr("concat('bronze_', table_name)")
      )
      .withColumn(
          "silver_table",
          expr("concat('silver_', table_name)")
      )
      .withColumn(
          "gold_table",
          expr("concat('gold_', table_name)")
      )
      .withColumn(
          "generated_on",
          current_timestamp()
      )
)
lineage_df.display()
```

```python
lineage_df.write \
    .mode("overwrite") \
    .saveAsTable(
        "accelerator.metadata.pipeline_lineage"
    )
```
