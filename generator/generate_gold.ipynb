```python
df = spark.table("accelerator.metadata.table_configs")
df.display()
```

```python
dimension_template = """
import dlt

@dlt.table(
    name="gold_{table_name}"
)
def gold_{table_name}():

    return (
        dlt.read("{silver_table}")
    )
"""
```

```python
for row in df.collect():
    
    tableName = row["table_name"]
    print(tableName)
    if row["gold_required"] == "Y":
        code = dimension_template.format(
            table_name=row["table_name"],
            silver_table=f"silver_{row['table_name']}"
        )
    
    # print(code)
    file_path = f"/Volumes/accelerator/metadata/generated_code/gold_{tableName}.py"

    with open(file_path, "w") as f:
        f.write(code)

    print(f"Generated: {file_path}")
```
