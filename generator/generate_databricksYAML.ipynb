```python
env_df = spark.table(
    "accelerator.metadata.environment_configs"
)

display(env_df)
```

```python
targets_yaml = ""

for row in env_df.collect():

    env = row["environment"]

    targets_yaml += f"""
  {env}:
"""

    if row["is_default"] == "Y":

        targets_yaml += """
    default: true
"""
```

```python
dab_yaml = f"""
bundle:
  name: accelerator_bundle

include:
  - resources/*.yml

targets:
{targets_yaml}
"""
```

```python
print(dab_yaml)
```

```python
databriciks_path = "/Volumes/accelerator/metadata/generated_code/bundle/databricks.yml"

with open(databriciks_path, "w") as f:
    f.write(dab_yaml)

print("databricks.yml generated")
```
