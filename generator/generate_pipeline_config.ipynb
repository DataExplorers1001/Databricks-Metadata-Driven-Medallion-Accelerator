```python
table_df = spark.read.table(
    "accelerator.metadata.table_configs"
)

pipeline_df = spark.read.table(
    "accelerator.metadata.pipeline_configs"
)
```

```python
display(table_df)
display(pipeline_df)
```

```python
pipeline = pipeline_df.first()

pipeline_name = pipeline["pipeline_name"]
catalog_name = pipeline["catalog_name"]
target_schema = pipeline["target_schema"]

print(pipeline_name)
print(catalog_name)
print(target_schema)
```

```python
import json

tables = []

for row in table_df.collect():

    table_name = row["table_name"]

    tables.append({
        "bronze_table": f"bronze_{table_name}",
        "silver_table": f"silver_{table_name}",
        "gold_table": f"gold_{table_name}"
    })

pipeline_json = {
    "pipeline_name": pipeline_name,
    "catalog": catalog_name,
    "target_schema": target_schema,
    "tables": tables
}
```

```python
print(
    json.dumps(
        pipeline_json,
        indent=4
    )
)
```

```python
file_path = "/Volumes/accelerator/metadata/generated_code/customer_pipeline.json"

with open(file_path, "w") as f:
    json.dump(
        pipeline_json,
        f,
        indent=4
    )

print("Pipeline JSON generated")
```
