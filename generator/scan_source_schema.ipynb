```python
table_df = spark.table(
    "accelerator.metadata.table_configs"
)

display(table_df)
```

```python
for row in table_df.collect():
    print(row)
    table_name = row["source_path"]
    table_name = table_name.replace("/Volumes/","")
    table_name = table_name.replace( "/",".")

    print(table_name)
    source_df = spark.table(table_name)
    print(source_df)

    for field in source_df.schema.fields:

        print(
            field.name,
            field.dataType,
            field.nullable
        )
```
