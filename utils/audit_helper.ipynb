```python
from pyspark.sql import Row
from datetime import datetime

def log_audit(
    pipeline_name,
    table_name,
    layer,
    status,
    records_processed=0
):

    audit_data = [
        (
            str(datetime.now().timestamp()),
            pipeline_name,
            table_name,
            layer,
            status,
            records_processed,
            datetime.now(),
            datetime.now(),
            0
        )
    ]

    spark.createDataFrame(
        audit_data,
        [
            "run_id",
            "pipeline_name",
            "table_name",
            "layer",
            "status",
            "records_processed",
            "start_time",
            "end_time",
            "duration_seconds"
        ]
    ).write.mode("append").saveAsTable(
        "accelerator.metadata.pipeline_run_audit"
    )
```
